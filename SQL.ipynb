```python
import sqlite3

conn = sqlite3.connect(":memory:")
cursor = conn.cursor()

```

```python
cursor.execute("""
CREATE TABLE students (
    id INTEGER,
    name TEXT,
    marks INTEGER
)
""")
```

```python
cursor.execute("""
INSERT INTO students VALUES
(1, 'Rohan', 85),
(2, 'Aman', 72),
(3, 'Priya', 91)
""")

cursor.execute("SELECT * FROM students")
cursor.fetchall()
```
