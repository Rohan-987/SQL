DataBase

A database is an organized collection of data that can be stored, accessed, and managed efficiently.

Types of Databases
Relational Database (RDBMS): Data stored in tables. Example: MySQL, PostgreSQL, Oracle.
NoSQL Database: Data stored as documents, key-value pairs, graphs, etc. Example: MongoDB.
Relational Database

Data is organized into tables consisting of:

Rows → individual records
Columns → attributes/fields

SQL is the language commonly used to create, retrieve, update, and delete data in relational databases.


Databases are costly. Cheaper alternatives are files.

Limitations in flat files:

A flat file stores data in a single file, such as CSV or TXT.
Data redundancy: Same data may be stored multiple times.
Data inconsistency: Duplicate data can become different in different files.
Limited relationships: Difficult to connect related data.
Poor security: Limited control over who can access specific data.
Difficult searching: Complex queries are harder to perform.
No proper data integrity: Hard to enforce rules such as unique IDs.
Scalability issues: Performance decreases as data grows.
Concurrency problems: Multiple users accessing/updating the file can cause conflicts.

DBMS
A DBMS is software used to create, store, manage, retrieve, and manipulate data in a database.

Examples: MySQL, PostgreSQL, Oracle, SQL Server, MongoDB.

Main Functions
Data storage – stores data efficiently.
Data retrieval – allows users to query data.
Data manipulation – INSERT, UPDATE, DELETE.
Security – controls user access.
Data integrity – maintains accurate and valid data.
Backup & recovery – protects data from loss.
Concurrency control – allows multiple users to access data safely.

DBMS vs Database

Database = the actual collection of data.
DBMS = software used to manage that data.
Example:
MySQL (DBMS) → manages → Employee Database


Functions of DBMS
Data Storage – Stores and organizes data efficiently.
Data Retrieval – Allows users to search and retrieve required data.
Data Manipulation – Supports INSERT, UPDATE, and DELETE.
Data Security – Controls access to the database.
Data Integrity – Ensures data remains accurate and consistent.
Backup & Recovery – Protects data and restores it after failures.
Concurrency Control – Allows multiple users to access the database safely.
Transaction Management – Ensures database operations are completed correctly.
Data Independence – Allows changes to the database structure without greatly affecting applications.


Types of DBMS
1.Hierarchical DBMS
Data is organized in a tree-like structure.
Example: IBM IMS.
2.Network DBMS
Data is organized as a network of records.
Supports many-to-many relationships.
Example: IDMS.
3.Relational DBMS (RDBMS)
Data is stored in tables (rows and columns).
Uses SQL to manage data.
Examples: MySQL, PostgreSQL, Oracle, SQL Server.
4.Object-Oriented DBMS (OODBMS)
Data is stored as objects, similar to object-oriented programming.
Example: ObjectDB.

DataBase Keys

Keys are attributes (columns) used to identify records and establish relationships between tables.

1.Primary Key
Uniquely identifies each record.
Cannot contain NULL values.
Example: Student_ID
2.Foreign Key
Connects two tables.
Refers to the primary key of another table.
Can contain duplicate values.
3.Candidate Key
A column or combination of columns that can uniquely identify a record.
One candidate key is chosen as the primary key.
4.Super Key
Any column or combination of columns that uniquely identifies a record.
May contain extra/unnecessary columns.
5.Alternate Key
A candidate key that was not chosen as the primary key.
6.Composite Key
A key made using two or more columns to uniquely identify a record.

Quick example:

Student(Student_ID, Email, Name, Course_ID)

Student_ID → Primary Key
Email → Candidate/Alternate Key
Course_ID → Foreign Key
(Student_ID, Course_ID) → Composite Key
