Order of Growth

What is Order of Growth?
Order of Growth tells us how the running time or memory usage of an algorithm increases as the input size (n) grows.
Instead of measuring time in seconds, we measure how quickly the work grows.
Example:
If an algorithm takes:
10 operations for 10 elements
100 operations for 100 elements
1000 operations for 1000 elements
Its growth is linear, written as:
O(n)

We compare algorithms mathematically. 

*Big-O Notation
 Big-O represents the upper bound (worst-case growth) of an algorithm. Denoted by Big O symbol.--O()

How to find the Big O for a problem example:
n^2*2n+2 ------> n^2 - square means nested loop
                 2n  - 2n is the number of operations going inside the loop. here 2 operations happen inside loop
                 2   - here 2 operations are happening outside all loops

Now n^2+n is left after we remove additive constants and multiplicative ones.
Rule* - We will consider the largest factors in the run time. So we ignore the n
 
The time complexity of this problem is O(n^2).The relation between time and input is Quadtratic.

Types of Order of Growth:
1.Constant
2.Linear
3.Quadratic
4.Logarithmic
5.n log n
6.exponential

Rank complexities from most to least efficient. The expected order is:
O(1) < O(logn) < O(n) < O(nlogn) < O(n^2) < O(n^3) < O(2n) < O(n!)

| Order of Growth | Complexity     | Example               |
| --------------- | -------------- | --------------------- |
| Constant        | **O(1)**       | Array indexing        |
| Logarithmic     | **O(log n)**   | Binary Search         |
| Linear          | **O(n)**       | Linear Search         |
| Linearithmic    | **O(n log n)** | Merge Sort, Heap Sort |
| Quadratic       | **O(n²)**      | Bubble Sort           |
| Cubic           | **O(n³)**      | Triple nested loops   |
| Exponential     | **O(2ⁿ)**      | Subset generation     |
| Factorial       | **O(n!)**      | Permutations          |

A good way to understand time complexity is to see how the input size (n) affects the number of operations (time).
# Time Complexity vs Input Size

| Complexity | Relationship Between Time and Input | If Input Doubles | Example (n = 10) | Example (n = 100) | Example (n = 1000) |
|------------|-------------------------------------|------------------|-----------------:|------------------:|-------------------:|
| **O(1)** | Time remains constant | No change | 1 | 1 | 1 |
| **O(log n)** | Time increases logarithmically | Increases slightly | ~3 | ~7 | ~10 |
| **O(n)** | Time increases linearly | Time doubles | 10 | 100 | 1000 |
| **O(n log n)** | Slightly faster than linear | About 2.3× | ~33 | ~664 | ~9966 |
| **O(n²)** | Time grows as the square of input | 4× increase | 100 | 10,000 | 1,000,000 |
| **O(n³)** | Time grows as the cube of input | 8× increase | 1,000 | 1,000,000 | 1,000,000,000 |
| **O(2ⁿ)** | Time doubles for every extra input | Extremely fast growth | 1,024 | ~1.27 × 10³⁰ | Impossible |
| **O(n!)** | Time grows factorially | Explodes rapidly | 3,628,800 | Astronomically large | Impossible |

A data structure is a way of organizing and storing data so that it can be accessed and modified efficiently.
It stores multiple items of same data type. It is a linear data structure.Memory location is continous.

Disadvantage
1.Fixed size (memory waste)
2.homogenous (lack of flexibility) - to fix this issue we use referntial arrays


Referential Array (Array of References)
A referential array is an array that stores references (memory addresses) to objects instead of storing the actual objects themselves. Python list is also known as referntial array.

Adv: can use multiple data types
Disadv: uses extra memory, is slightly slow bcoz it has to fetch address and then the data


Dynamic Array
A dynamic array is an array that automatically resizes itself when it becomes full, allowing elements to be added or removed dynamically.
A dynamic array is an array that can automatically grow (and sometimes shrink) in size during runtime.
Unlike a normal (static) array, you don't have to decide its size in advance.
Python list is a dynamic array.
